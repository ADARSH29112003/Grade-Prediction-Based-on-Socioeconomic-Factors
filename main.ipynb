```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
numerical_columns = df.select_dtypes(include = np.number).columns
print(numerical_columns)
```

```python
df[numerical_columns].isnull().sum()
```

```python
numerical_columns = ['annual_family_income_inr', 'distance_to_school_km', 'attendance_rate_pct', 'previous_year_grade_pct']
```

```python
df[numerical_columns].median()
```

```python
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
# Empty space is filled with median number in every column 
# For example: attendance_rate_pct column is it empty space 86.0 filled
df[numerical_columns].isnull().sum()
```

```python
categorial_columns = df.select_dtypes(include='str').columns
```

```python
df[categorial_columns].isnull().sum()
```

```python
df[categorial_columns].mode().iloc[0]
```

```python
for column in categorial_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
df[categorial_columns].isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
# has_internet_access -> 180 values are empty (Mode)
df['has_internet_access'].mode()[0]
```

```python
# Now we will fill the missing value in these column 'has_internet_access' 
df['has_internet_access'] = df['has_internet_access'].fillna(df['has_internet_access'].mode()[0])
```

```python
df.isnull().sum().sum()
```

```python
# Final Grade Distribution
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='final_grade_pct', bins=20, kde=True)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (%)')
plt.ylabel('Number of Students')

plt.show()
```

```python
# Attendance Rate Distribution
plt.figure(figsize=(8,5))

sns.histplot(data = df, x = 'attendance_rate_pct', bins = 20, kde = True)

plt.title('Distribution of Student Attendance')
plt.xlabel('Attendance Rate (%)')
plt.ylabel('Number of Students')

plt.show()
```

```python
# Study Hours Distribution
plt.figure(figsize=(8,5))

sns.histplot(data = df, x = 'study_hours_per_week', bins = 20, kde = True)

plt.title('Distribution of Weekly Study Hours')
plt.xlabel('Study Hours per Week')
plt.ylabel('Number of Students')

plt.show()
```

```python
# Student Age Distribution
plt.figure(figsize=(8,5))

sns.histplot(data = df, x = 'age', bins = 9, kde = True)

plt.title('Distribution of Student Age')
plt.xlabel('Age (Years)')
plt.ylabel('Number of Students')

plt.show()
```

```python
# 
plt.figure(figsize=(8,5))

sns.histplot(data = df, x = 'family_size', bins = 9, kde = True)

plt.title('Distribution of Family Size')
plt.xlabel('Number of Family Members')
plt.ylabel('Number of Students')

plt.show()
```

```python
# At-Risk Student Analysis

plt.Figure(figsize = (8, 5))

sns.countplot(data= df, x = 'at_risk_flag')

plt.title('At-Risk VS Non-At-Risk Students')
plt.xlabel('Student Risk Status')
plt.ylabel('Number of Students')

plt.xticks([0,1], ['Non-At-Risk', 'At-Risk'])

plt.show()
```
